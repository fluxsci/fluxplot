```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
fake_data1 = np.random.randint(10, 20, 100)
fake_data2 = np.random.randint(15, 30, 100)
```

```python
fake_ts = np.random.randint(24, 28, 200)
fake_time = np.arange(0, 200, 1)
```

```python
plt.rcdefaults()
plt.rcParams['font.size'] = 12
f, ax = plt.subplots(1, 1, figsize=(3, 3))
ax = ax.boxplot([fake_data1, fake_data2])
f.savefig('./examples/sizing/boxplot_3x3_dpi300.svg', dpi=300, bbox_inches='tight', transparent=True)
```

```python
plt.rcdefaults()
plt.rcParams['font.size'] = 12
f, ax = plt.subplots(1, 1, figsize=(7, 2))
ax.plot(fake_time, fake_ts, color='firebrick', linewidth=2)
ax.set_ylim(15, 35)
f.savefig('./examples/sizing/lineplot_7x2_NO-BBOX.svg', transparent=True)
```
